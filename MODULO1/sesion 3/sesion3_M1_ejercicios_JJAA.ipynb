```python
cartera = [

{'id':'A-101','nom':'Sofia','edad':24,'veh':'Sedan', 'siniest':0,'suma':300_000},

{'id':'A-102','nom':'Miguel','edad':38,'veh':'SUV', 'siniest':1,'suma':500_000},

{'id':'A-103','nom':'Laura', 'edad':31,'veh':'Hatchback','siniest':0,'suma':250_000},

{'id':'A-104','nom':'Roberto','edad':55,'veh':'SUV','siniest':2,'suma':600_000},

{'id':'A-105','nom':'Carmen','edad':42,'veh':'Sedan','siniest':0,'suma':350_000}]

tarifas = {'Sedan':0.025, 'SUV':0.035, 'Hatchback':0.022, 'Pick-up':0.040}
```

```python
# 1. Calcula la prima de cada poliza: suma_asegurada × tasa_tarifa[vehiculo]. Agrega el campo 'prima' al dict.
for p in cartera :
    p['prima'] = p['suma']*tarifas[p['veh']]
    print(f"Prima del asegurado {p['nom']} es de: ${p['prima']:,.2f}")
```

```python
# 2. Clasifica cada asegurado: 'Riesgo Alto' si siniest >= 2, 'Riesgo Medio' si siniest == 1, 'Bajo' si siniest == 0.

for p in cartera:
    if p['siniest'] >= 2:
        p['Riesgo'] = "Riesgo Alto"
    elif  p['siniest'] == 1:
        p['Riesgo'] = "Riesgo Medio"
    else:
        p['Riesgo'] = "Riesgo Bajo"
    print(f'Riesgo del asegurad@ {p["nom"]} es {p["Riesgo"]}') 
```

```python
# 3. Imprime un reporte tabular: ID | Nombre | Vehiculo | Prima | Clasificacion.

print("="*65)
print("Reporte tabular")
print("="*65)

print(f'{"ID":<8} {"Nombre":<14} {"Vehiculo":<12} {"Prima":<9}   {"Riesgo":<8}')

print('-' * 65)

for p in cartera:
    print(f"{p['id']:<8} {p['nom']:<14} {p['veh']:<12} ${p['prima']:<9,.2f}   {p['Riesgo']:<8}")
```

```python
# 4. Usando comprehensions: lista de primas de SUVs, promedio de primas, total de siniestros.

prima_SUVs = [p['prima'] for p in cartera if p['veh'] == "SUV"]
Siniestros_SUVs = [p['siniest'] for p in cartera if p['veh'] == "SUV"]
promedio_sv = sum(prima_SUVs)/ len(prima_SUVs)
tot_sin_SUVs = sum(Siniestros_SUVs)
 
print(f'Primas SUVs :{prima_SUVs}')
print(f'Prima promedio : ${promedio_sv:,.2f}')
print(f'Siniestros totales SUVs : {tot_sin_SUVs}')
```

```python
# 5. Bonus: ordena la cartera por prima descendente e imprime el ranking.

ranking = sorted(cartera, key= lambda p:p['prima'], reverse=True)

for i,p in enumerate(ranking,1):
     print(f'{i:<5} {p["nom"]:<14} {p["veh"]:<12} ${p["prima"]:<8,.0f}')
```
